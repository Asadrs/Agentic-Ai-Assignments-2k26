```python
# ==============================
# Assignment-03
# Few-Shot + Chain-of-Thought + Persona Document
# ==============================

from openai import OpenAI

client = OpenAI(
    api_key="api key here",          # paste your xai- key here
    base_url="https://api.x.ai/v1"
)
print("API client ready!\n")


# ==========================================
# Question 1: Few-Shot Sentiment Classification
# ==========================================

print("="*50)
print("Question 1: Few-Shot Sentiment Classification")
print("="*50)

few_shot_prompt = """
Classify the customer message into Positive, Negative, or Neutral.

Examples:

Message: "I love this product, it works perfectly!"
Sentiment: Positive

Message: "The delivery was late and the item was damaged."
Sentiment: Negative

Message: "The product is okay, nothing special."
Sentiment: Neutral

Message: "Customer support was very helpful and resolved my issue quickly."
Sentiment: Positive

Now classify this message:
Message: "The app keeps crashing every time I try to login."
Sentiment:
"""

response1 = client.chat.completions.create(
    model="grok-3",
    messages=[{"role": "user", "content": few_shot_prompt}],
    temperature=0
)

print("Predicted Sentiment:", response1.choices[0].message.content.strip())
print()


# ==========================================
# Question 2: Chain-of-Thought Math Problem
# ==========================================

print("="*50)
print("Question 2: Chain-of-Thought Prompting")
print("="*50)

problem = """
A shopkeeper bought 15 boxes of apples. Each box contains 24 apples.
He sold 180 apples and then found that 12 apples were rotten.
How many good apples does he still have?
"""

cot_prompt = f"""
Solve the following problem step by step. Show your reasoning clearly.
At the end, write the final answer clearly.

Problem:
{problem}

Let's think step by step:
"""

response2 = client.chat.completions.create(
    model="grok-3",
    messages=[{"role": "user", "content": cot_prompt}],
    temperature=0
)

print(response2.choices[0].message.content)
print()


# ==========================================
# Question 3: Persona + Document Grounding
# ==========================================

print("="*50)
print("Question 3: Persona + Document")
print("="*50)

document = """
Python is a high-level programming language created by Guido van Rossum.
It was first released in 1991. Python supports multiple programming paradigms 
including procedural, object-oriented, and functional programming.
The official package manager for Python is called pip.
Python uses indentation to define code blocks instead of curly braces.
"""

system_prompt = f"""
You are a helpful Python Tutor.
You must answer questions ONLY using the information given in the document below.
If the answer is not present in the document, reply exactly with:
"I don't know based on the provided document."

Document:
{document}
"""

def ask_question(question):
    response = client.chat.completions.create(
        model="grok-3",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
        ],
        temperature=0
    )
    return response.choices[0].message.content.strip()

print("Question 1: Who created Python?")
print("Answer:", ask_question("Who created Python?"))
print()

print("Question 2: What is the capital of France?")
print("Answer:", ask_question("What is the capital of France?"))
```
